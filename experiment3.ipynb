```python
#change given string to new string
s=input("enter a string:")
new_string=s[-1]+s[1:-1]+s[0]
print("new string:",new_string)
```

```python
#get a string made of the first 2 and last 2 chars from a given string
s=input("enter a string:")
new_string=s[:2]+s[-2:]
print("new string:",new_string)

```

```python
#single string separated with space from two strings by swapping the character at position 1
s1=input("enter first string:")
s2=input("enter second string:")
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print(new_s1 + " " + new_s2)
```

```python
#get a string from a given string where all occurences of its first char have been changed to $,except the first char itself
s=input("enter a string:")
first=s[0]
new_string=first+s[1:].replace(first, "$")
print("new string:",new_string)
```
