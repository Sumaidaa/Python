```python
#to get the third elementand third element from the last of a tuple
t=tuple(input("enter the values separated by comma:").split(","))
print("3rd element:",t[2])
print("3rd element from last:",t[-3])

```

```python
#program which accepts a sequence of comma separated numbers from user and genarate a tuple with those numbers and find average of all items.
numbers=input("enter number separated by comma:")
t=tuple(map(int,numbers.split(",")))
average=sum(t)/len(t)
print("tuple:",t)
print("average:",average)
```
