```python
#program that accepts an integer and compute the value of n+nn+nnn
n=input("enter a number:")
n1=int(n)
n2=int(n*2)
n3=int(n*3)
result=n1+n2+n3
print("result",n1,"+",n2,"+",n3,"=",result)
```

```python
#area of a circle
import math
r=float(input("enter a number:"))
area=math.pi*r*r
print(f"{area:.2f}")
```

```python
#biggest of three numbers using max
a=int(input("enter first number:"))
b=int(input("enter second number:"))
c=int(input("enter third number:"))
biggest=max(a,b,c)
print("biggest number=",biggest)
```
