```python
#python program to printout all colors from colorset 1 not contained in color set 2
color_set1=set(input("enter colors in set 1:").split(","))
color_set2=set(input("enter colors in set 2:").split(","))
print("set 1:",color_set1)
print("set 2:",color_set2)
result=color_set1-color_set2
print("colors in set 1 but not in set 2:")
print(result)
```

```python
#program to print all colors from color set 1 and color set 2
color_set1=set(input("enter colors in set 1:").split(","))
color_set2=set(input("enter colors in set 2:").split(","))
print("set 1:",color_set1)
print("set 2:",color_set2)
result=color_set1&color_set2
print("colors present in both sets:")
print(result)
```
