```python
#program to accept 5 students names&roll numbers and map these into a dictionary.sort the dictionary in ascending and descending order of their roll number
students={}
roll1=int(input("enter roll no 1:"))
name1=input("enter name 1:")
students[roll1]=name1
roll2=int(input("enter roll no 2:"))
name2=input("enter name 2:")
students[roll2]=name2
roll3=int(input("enter roll no 3:"))
name3=input("enter name 3:")
students[roll3]=name3
roll4=int(input("enter roll no 4:"))
name4=input("enter name 4:")
students[roll4]=name4
roll5=int(input("enter roll no 5:"))
name5=input("enter name 5:")
students[roll5]=name5
print("original dictionary:",students)
print("ascending order:",dict(sorted(students.items())))
print("descending order:",dict(sorted(students.items(),reverse=True)))
```
